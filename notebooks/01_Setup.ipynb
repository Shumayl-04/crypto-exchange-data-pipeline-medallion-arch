### Create External Location

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS cryptoexchangedata_crypto_exchange_container
    URL 'abfss://crypto-exchange-container@cryptoexchange.dfs.core.windows.net/' 
    WITH (STORAGE CREDENTIAL crypto_exchange_ext_sc)
    COMMENT 'External location for crypto exchange data'
```

### Create Catalog - ecommerce

```python
%sql
SHOW CATALOGS;
```

```python
%sql
CREATE CATALOG IF NOT EXISTS crypto_exchange
    MANAGED LOCATION 'abfss://crypto-exchange-container@cryptoexchange.dfs.core.windows.net/'
    COMMENT 'Catalog for crypto exchange data';
```

### Create Schema
1. landing
2. Bronze
3. Silver
4. Gold

```python
%sql
SELECT current_catalog();
```

```python
%sql
USE CATALOG crypto_exchange;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS landing
    MANAGED LOCATION 'abfss://crypto-exchange-container@cryptoexchange.dfs.core.windows.net/landing';

CREATE SCHEMA IF NOT EXISTS bronze
    MANAGED LOCATION 'abfss://crypto-exchange-container@cryptoexchange.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS silver
    MANAGED LOCATION 'abfss://crypto-exchange-container@cryptoexchange.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS gold
    MANAGED LOCATION 'abfss://crypto-exchange-container@cryptoexchange.dfs.core.windows.net/gold';
```

```python
%sql
SHOW SCHEMAS;
```

## Create Volume

```python
%sql
USE CATALOG crypto_exchange;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS cryptoexchange_data
    LOCATION 'abfss://crypto-exchange-container@cryptoexchange.dfs.core.windows.net/landing/cryptoexchange_data';
```

```python
%fs ls /Volumes/crypto_exchange/landing/cryptoexchange_data
```
